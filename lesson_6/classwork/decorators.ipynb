```python
def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper
```

```python
@decorator
def say_hello():
    print("Hello")

say_hello()
```

```python
def bold(func):
    def wrapper():
        return("<b>" + func() + "</b>")
    return wrapper

@bold
def greet():
    return "Hello"

print(greet())
```

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def greet(name):
    print(f"Hello {name}")

greet("Alisa")
```

```python
def my_func(func):
    def wrapper(*args, **kwargs):
            return func(*args, **kwargs)
    return wrapper  

@my_func
def greet(name):
    print(f"Welcome, {name}")

greet("Alisa")
```

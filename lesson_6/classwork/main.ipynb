```python
def counter():
    print("Boshlandi")
    yield 1
    print("Davom etdi")
    yield 2
    print("Tugadi")
    yield 3

gen = counter()
next(gen)  # "Boshlandi" chop etadi, 1 qaytaradi
next(gen)  # "Davom etdi" chop etadi, 2 qaytaradi
next(gen)  # "Tugadi" chop etadi, 3 qaytaradi
```

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1
    

for x in countdown(5):
    print(x)
# 5
# 4
# 3
# 2
# 1
# Poyga boshlandi!
```

```python
def juft_sonlar():
    n = 0
    while True:
        yield n
        n += 2

gen = juft_sonlar()

for _ in range(6):
    print(next(gen))
```

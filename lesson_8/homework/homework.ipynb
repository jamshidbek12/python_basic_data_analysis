```python
class Account():
    def __init__(self, account_number, name, balance=0 ):
        self.account_number = account_number
        self.name = name
        self.balance = balance

    def __str__(self):
        return f"Account({self.account_number}, {self.name}, balance={self.balance})"

class Bank():
    def __init__(self):
        self.accounts = {}
        self.next_account_number = 1001

    def create(self, name, initial_deposit):
        account_number = str(self.next_account_number)
        new_account = Account(account_number, name, initial_deposit)
        self.accounts[account_number] = new_account
        self.next_account_number += 1
        return self.accounts

user1 = Bank()
print(user1.create("John", 500))

for acc_num, acc_obj in user1.accounts.items():
    print(f"{acc_num}: {acc_obj.name}, balance={acc_obj.balance}")



```

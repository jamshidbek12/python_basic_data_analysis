```python
import numpy as np
```

```python
# Task 1
temperatures = [32, 68, 100, 212, 77]
arr_temps = (np.array(temperatures)).astype(np.int64)

@np.vectorize
def fah_to_cel(fah):
    # res_temps = (fah - 32) * (5 / 9)
    return (fah - 32) * 5 / 9

print(fah_to_cel(arr_temps))
```

```python
# Task 2
@np.vectorize
def power_func(base, power):
    return base ** power

bases = np.array([2,3,4,5])
powers = np.array([1,2,3,4])

print(power_func(bases, powers))
```

```python
# Task 3
W = np.array([
    [4, 5, 6],
    [3, -1, 1],
    [2, 1, -2]
])

Y = np.array([7, 4, 5])
print(np.linalg.solve(W, Y))
```

```python
# Task 4
W = np.array([
    [10, -2, 3],
    [-2, 8, -1],
    [3, -1, 6]
])

Y = np.array([12, -5, 15])

print(np.linalg.solve(W, Y))
```

```python
# Image Manipulation with NumPy and PIL
from PIL import Image

with Image.open('images/bird.jpg') as img:
    print(img.mode)
    img_arr = np.array(img)
    print(img_arr.shape)

def flip_image(flipping_img):
    flipped = flipping_img[::-1,::-1]
    res_img = Image.fromarray(flipped)
    res_img.save('images/flip.jpg')
    return res_img

def add_noise(noising_img):
    gray_img_arr = (noising_img[:, :, 0] * 0.299 + noising_img[:, :, 1] * 0.587 + noising_img[:, :, 2] * 0.114).astype(np.int8)
    res = Image.fromarray(gray_img_arr, mode='L')
    res.save('images/noised.jpg')
    return res

def brighten_channels(bright_img):
    img = bright_img.astype(int)
    img[:,:,0] += 40
    img = np.clip(img, 0, 255)

    res = Image.fromarray(img.astype(np.uint8))
    res.save('images/brightened.jpg')
    return res

def apply_mask(img_black):
    img = img_black.astype(int)
    img[400:500,700:800] = 0
    res = Image.fromarray(img.astype(np.uint8))
    res.save('images/black_masked.jpg')
    return res

flip_image(img_arr)
add_noise(img_arr)
brighten_channels(img_arr)
apply_mask(img_arr)
```

```python
x = [
    [1,2],
    [3,4]
]
y = x.copy()
x[0].append(5)

print(x)
print(y)
print(id(x))
print(id(y))
```

```python
from copy import deepcopy, copy

x = [
    [1,2],
    [3,4]
]
y = deepcopy(x)
x[0].append(5)

print(x)
print(y)
print(id(x))
print(id(y))
```
